## 2) Початкова дескриптивна статистика: числові змінні

```python
numeric_cols = ["Виставлена вартість","Дохід","Кількість пропозицій","Вкотре виставляється","Мінімальна кількість учасників",
    "Експертна грошова оцінка","Нормативна грошова оцінка","Площа земельної ділянки, ГА","Сума найвищої цінової пропозиції",
    "price_lease_ha","price_change","price_change%"]

def numeric_summary(df, columns):
    rows = []

    for col in columns:
        s = df[col].dropna()

        q1 = s.quantile(0.25)
        q3 = s.quantile(0.75)
        iqr = q3 - q1

        lower_bound = q1 - 1.5 * iqr
        upper_bound = q3 + 1.5 * iqr

        outliers_count = ((s < lower_bound) | (s > upper_bound)).sum()

        rows.append({
            "column": col,
            "count": s.count(),
            "missing": df[col].isna().sum(),
            "mean": s.mean(),
            "median": s.median(),
            "std": s.std(),
            "min": s.min(),
            "Q1": q1,
            "Q3": q3,
            "max": s.max(),
            "IQR": iqr,
            "outliers_count": outliers_count,
            "outliers_%": outliers_count / len(s) * 100
        })

    return pd.DataFrame(rows)

desc_stat_1 = numeric_summary(df,numeric_cols)
desc_stat_1
```

## Початкова дескриптивна статистика: категоріальні змінні

```python
categorical_cols = ["Статус аукціону","Організатор","Переможець","Тривалість оренди","КВЦПЗ опис класифікатора товару","Тип власності",
"Регіон місцезнаходження","Місто місцезнаходження","область","oblast_kadastr","Наявність обтяжень","Наявність співвласників"]
def categorical_summary(df, columns):
    rows = []

    for col in columns:
        vc = df[col].value_counts(dropna=False)

        rows.append({
            "column": col,
            "unique_count": df[col].nunique(dropna=True),
            "missing_count": df[col].isna().sum(),
            "missing_%": df[col].isna().mean() * 100,
            "most_common": vc.index[0],
            "most_common_count": vc.iloc[0],
            "most_common_%": vc.iloc[0] / len(df) * 100
        })

    return pd.DataFrame(rows)

categorical_stat_1 =categorical_summary(df, categorical_cols)
categorical_stat_1
```

```python
continuous_cols = [
    "Виставлена вартість",
    "Дохід",
    "Нормативна грошова оцінка",
    "Площа земельної ділянки, ГА",
    "Сума найвищої цінової пропозиції",
    "price_lease_ha",
    "price_change",
    "price_change%"
]
def iqr_outliers(df, columns):
    rows = []
    iqr_flags = pd.DataFrame(index=df.index)
    for col in columns:
        s = df[col].dropna()

        q1 = s.quantile(0.25)
        q3 = s.quantile(0.75)
        iqr = q3 - q1

        lower = q1 - 1.5 * iqr
        upper = q3 + 1.5 * iqr

        mask = (df[col] < lower) | (df[col] > upper)
        iqr_flags[col] = mask
        rows.append({
            "column": col,
            "lower_bound": lower,
            "upper_bound": upper,
            "outliers_count": mask.sum(),
            "outliers_%": round(mask.mean() * 100, 2)
        })

    return pd.DataFrame(rows), iqr_flags
iqr_summary, iqr_flags = iqr_outliers(df, continuous_cols)
iqr_summary
```

```python
def mad_outliers(series, threshold=3.5):
    s = series.dropna()

    median = s.median()
    mad = (s - median).abs().median()

    if mad == 0:
        return pd.Series(False, index=series.index)

    modified_z = 0.6745 * (series - median) / mad

    return modified_z.abs() > threshold

mad_summary = []
mad_flags = pd.DataFrame(index=df.index)
for col in continuous_cols:
    mask = mad_outliers(df[col])
    mad_flags[col] = mask
    mad_summary.append({
        "column": col,
        "outliers_count": mask.sum(),
        "outliers_%": round(mask.mean() * 100, 2)
    })

mad_summary = pd.DataFrame(mad_summary)
mad_summary
```

## Підсумок перевірки аномалій

Під час перевірки даних було виявлено кілька груп потенційних проблем, які потребували окремої валідації перед очищенням.

1. **Некоректні дати договорів.** У 25 записах `"Дата підписання договорів"` була раніше за `"Дата аукціону"`. Такі значення не можуть бути коректними та потребують виправлення.

2. **Нетипово великі інтервали до підписання договору.** У `days_to_contract` є дуже великі значення, зокрема інтервали у кілька років. Такі записи були перевірені на можливі помилки в датах.

3. **Розбіжність між `"Дохід"` і `"Сума найвищої цінової пропозиції"`.** Приблизно у 3 000 записах ці поля відрізняються. Сама розбіжність не означає помилку, тому для визначення фактичної суми переможця була потрібна додаткова перевірка через Prozorro API.

4. **Регіональні невідповідності.** У частині записів значення `"область"` і `oblast_kadastr` не збігаються, тому для аналізу потрібно було обрати єдине джерело регіональної інформації.

5. **Повторні кадастрові номери.** 385 кадастрових номерів зустрічаються більш ніж в одному аукціоні та охоплюють 783 записи. Вони не розглядалися як звичайні дублікати, оскільки можуть відповідати повторному виставленню тієї самої ділянки.

6. **Експертна грошова оцінка.** Майже всі значення у `"Експертна грошова оцінка"` дорівнюють 0, тому було перевірено, чи означає 0 реальне значення, чи відсутність / незастосовність показника.

7. **Потенційні статистичні викиди.** Методи IQR і MAD виявили значну кількість екстремальних значень у цінових показниках. Такі спостереження не видалялися автоматично, оскільки частина з них може бути реальним результатом аукціону.

8. **Аномальні значення площі.** У `"Площа земельної ділянки, ГА"` були як надзвичайно великі значення понад 100 000 га, так і дуже малі значення близько 0,0001 га. Це могло суттєво викривляти розрахунок `price_lease_ha`, тому площі потребували окремої перевірки.

9. **Показник `price_change%`.** Перевірка формули показала, що поле збережене як відношення, а не як звичайний відсоток. Для коректної інтерпретації показник потрібно було перерахувати.

Основний принцип подальшого очищення: **екстремальні значення не вважалися помилковими лише через їхню величину. Корекція виконувалася тільки після додаткової перевірки.**

### Валідація даних через Prozorro API

Код нижче використовувався для додаткової перевірки підозрілих значень площі земельних ділянок і доходу. Основний робочий процес використовує вже очищені значення, тому для повторного запуску аналізу live-запити до API не потрібні.

```python
import requests
import re
from concurrent.futures import ThreadPoolExecutor

def get_area_from_api(auction_id):
    try:
        data = requests.get(
            f"https://procedure.prozorro.sale/api/search/procedures/byAuctionId/{auction_id}",
            timeout=10
        ).json()

        texts = [
            data.get("description", {}).get("uk_UA", ""),
            data.get("title", {}).get("uk_UA", "")
        ]

        for item in data.get("items", []):
            texts.append(
                item.get("description", {}).get("uk_UA", "")
            )

        text = " ".join(texts)

        m = re.search(
            r'(?:площею|площа|загальною площею)\s*[:\-]?\s*(\d+(?:[.,]\d+)?)\s*га',
            text,
            re.I
        )

        return float(m.group(1).replace(",", ".")) if m else None

    except:
        return None
```

```python
# API validation is intentionally disabled for the reviewer workflow.
RUN_API_VALIDATION = False
if RUN_API_VALIDATION:
    with ThreadPoolExecutor(max_workers=10) as executor:
        df_clean["area_from_api"] = list(executor.map(get_area_from_api, df_clean["Ідентифікатор аукціону"]))
else:
    df_clean["area_from_api"] = np.nan
```

```python
def get_real_income(auction_id):
    try:
        data = requests.get(
            f"https://procedure.prozorro.sale/api/search/procedures/byAuctionId/{auction_id}",
            timeout=10
        ).json()

        for a in data.get("awards", []):
            if a.get("status") in ["active", "pending"]:
                return a.get("value", {}).get("amount")

        return None

    except:
        return None
```

```python
# API validation is intentionally disabled for the reviewer workflow.
if RUN_API_VALIDATION:
    with ThreadPoolExecutor(max_workers=10) as executor:
        df_clean["real_income_api"] = list(executor.map(get_real_income, df_clean["Ідентифікатор аукціону"]))
else:
    df_clean["real_income_api"] = np.nan
```

```python
df_clean["area_original"] = df["Площа земельної ділянки, ГА"]
df_clean["Площа земельної ділянки, ГА"] = (df_clean["area_from_api"].combine_first(df_clean["area_original"]))
df_clean["Дохід_original"] = df["Дохід"]
df_clean["Дохід"] = (df_clean["real_income_api"].combine_first(df_clean["Дохід"]))
```

### Тести Краскела—Волліса

```python
from scipy.stats import kruskal

price_growth_groups = [competition_data.loc[competition_data["competition_group"] == group, "price_change_pct"]
                       for group in ["1", "2", "3", "4", "5+"]]
price_ha_groups = [competition_data.loc[competition_data["competition_group"] == group, "price_lease_ha"]
                   for group in ["1", "2", "3", "4", "5+"]]

price_growth_test = kruskal(*price_growth_groups)
price_ha_test = kruskal(*price_ha_groups)

price_growth_p = f"= {price_growth_test.pvalue:.3g}" if price_growth_test.pvalue > 0 else "< 0.001"
price_ha_p = f"= {price_ha_test.pvalue:.3g}" if price_ha_test.pvalue > 0 else "< 0.001"

print(f"Price growth: statistic = {price_growth_test.statistic:.2f}, p-value {price_growth_p}")
print(f"Price per ha: statistic = {price_ha_test.statistic:.2f}, p-value {price_ha_p}")
```

Тести Краскела—Волліса статистично значущі як для зростання ціни (H = 10 240,86; p < 0,001), так і для річної ціни оренди за гектар (H = 8 829,50; p < 0,001).

Це означає, що розподіли обох показників статистично відрізняються між групами з різною кількістю ставок.

### Прості OLS-моделі

```python
import statsmodels.formula.api as smf

price_growth_data = df_clean[["Кількість пропозицій", "price_change_pct"]].dropna()
model_price_growth = smf.ols('price_change_pct ~ Q("Кількість пропозицій")', data=price_growth_data).fit()

price_ha_data = df_clean[["Кількість пропозицій", "price_lease_ha"]].dropna()
price_ha_data = price_ha_data[price_ha_data["price_lease_ha"] > 0].copy()
price_ha_data["log_price_lease_ha"] = np.log(price_ha_data["price_lease_ha"])
model_price_ha = smf.ols('log_price_lease_ha ~ Q("Кількість пропозицій")', data=price_ha_data).fit()

ols_summary = pd.DataFrame({
    "bids_coefficient": [model_price_growth.params.iloc[1], model_price_ha.params.iloc[1]],
    "p_value": [model_price_growth.pvalues.iloc[1], model_price_ha.pvalues.iloc[1]],
    "R_squared": [model_price_growth.rsquared, model_price_ha.rsquared]
}, index=["price_change_pct", "log(price_lease_ha)"])
ols_summary["p_value"] = ols_summary["p_value"].map(lambda p: f"{p:.3e}")
ols_summary.round(4)
```

У простій OLS-моделі для зростання ціни коефіцієнт кількості ставок є статистично значущим, однак модель має дуже низьку пояснювальну здатність (R² ≈ 0,014).

Тому сам коефіцієнт не варто переоцінювати: кількість ставок пов’язана зі зростанням ціни, але сама по собі пояснює лише невелику частину його варіації.

Для опису цього зв’язку кореляція Спірмена та порівняння медіан між групами є більш інформативними.

### OLS-модель з контролями

```python
control_data = df_clean[["Кількість пропозицій", "price_lease_ha", "Площа земельної ділянки, ГА", "Виставлена вартість"]].dropna()
control_data = control_data[(control_data["price_lease_ha"] > 0) &
                            (control_data["Площа земельної ділянки, ГА"] > 0) &
                            (control_data["Виставлена вартість"] > 0)].copy()

control_data["log_price_lease_ha"] = np.log(control_data["price_lease_ha"])
control_data["log_area"] = np.log(control_data["Площа земельної ділянки, ГА"])
control_data["log_start_price"] = np.log(control_data["Виставлена вартість"])

model_controls = smf.ols(
    'log_price_lease_ha ~ Q("Кількість пропозицій") + log_area + log_start_price',
    data=control_data
).fit()

control_results = pd.DataFrame({"coefficient": model_controls.params, "p_value": model_controls.pvalues})
print(f"R-squared: {model_controls.rsquared:.3f}")
control_results["p_value"] = control_results["p_value"].map(lambda p: f"{p:.3e}")
control_results.round(4)
```

Після контролю площі земельної ділянки та стартової ціни коефіцієнт кількості ставок залишається позитивним (0,257) і статистично значущим (p < 0,001).

Це відповідає приблизно 29,3% вищій річній ціні оренди за гектар на кожну додаткову ставку, за інших рівних умов у межах моделі.

Модель пояснює близько 47,9% варіації логарифма ціни оренди за гектар (R² = 0,479).

Позитивний зв’язок зберігається після базових контролів, однак результат слід трактувати як асоціацію, а не як причинний ефект.

### Bootstrap-довірчий інтервал

```python
np.random.seed(42)

one_bid = competition_data.loc[competition_data["competition_group"] == "1", "price_lease_ha"].to_numpy()
five_plus_bids = competition_data.loc[competition_data["competition_group"] == "5+", "price_lease_ha"].to_numpy()

observed_difference = np.median(five_plus_bids) - np.median(one_bid)
bootstrap_differences = []

for _ in range(1000):
    one_bid_sample = np.random.choice(one_bid, size=len(one_bid), replace=True)
    five_plus_sample = np.random.choice(five_plus_bids, size=len(five_plus_bids), replace=True)
    bootstrap_differences.append(np.median(five_plus_sample) - np.median(one_bid_sample))

confidence_interval = np.percentile(bootstrap_differences, [2.5, 97.5])
print(f"Observed median difference: {observed_difference:.2f} UAH per ha")
print(f"95% confidence interval: [{confidence_interval[0]:.2f}, {confidence_interval[1]:.2f}]")

plt.figure(figsize=(8, 5))
plt.hist(bootstrap_differences, bins=30, edgecolor="black")
plt.axvline(0, color="red", linestyle="--", label="Zero difference")
plt.title("Bootstrap median differences: 5+ bids vs 1 bid")
plt.xlabel("Median price per ha difference, UAH")
plt.ylabel("Frequency")
plt.legend()
plt.show()
```

The observed median price-per-hectare difference between 5+ bids and 1 bid is 15,459.52 UAH. The 95% bootstrap confidence interval is [14,950.87; 15,920.01] UAH, so it does not include zero. This supports a higher median price per hectare in auctions with 5+ bids.

```python
from scipy.stats import kruskal

area_price_groups = [area_data.loc[area_data["area_group"] == group, "price_lease_ha"]
                     for group in ["<1 ha", "1–5 ha", "5–10 ha", "10–50 ha", "50+ ha"]]
area_test = kruskal(*area_price_groups)

print(f"Price per ha: statistic = {area_test.statistic:.2f}, p-value = {area_test.pvalue:.3g}")
```

Тест Краскела—Волліса є статистично значущим (H = 262,11; p < 0,001), тому розподіли ціни оренди за гектар відрізняються між групами земельних ділянок різної площі.

Водночас статистична значущість не означає, що площа є сильним пояснювальним фактором: при великій кількості спостережень навіть невеликі відмінності можуть бути статистично значущими.

### Проста OLS-модель

```python
import statsmodels.formula.api as smf

area_ols_data = df_clean[["Площа земельної ділянки, ГА", "price_lease_ha"]].dropna()
area_ols_data = area_ols_data[(area_ols_data["Площа земельної ділянки, ГА"] > 0) &
                             (area_ols_data["price_lease_ha"] > 0)].copy()
area_ols_data["log_area"] = np.log(area_ols_data["Площа земельної ділянки, ГА"])
area_ols_data["log_price_lease_ha"] = np.log(area_ols_data["price_lease_ha"])

area_model = smf.ols("log_price_lease_ha ~ log_area", data=area_ols_data).fit()

area_ols_summary = pd.DataFrame({
    "area_coefficient": [area_model.params["log_area"]],
    "p_value": [area_model.pvalues["log_area"]],
    "R_squared": [area_model.rsquared]
}, index=["log_area"])
area_ols_summary
```

У простій log-log OLS-моделі коефіцієнт площі становить 0,063 і є статистично значущим.

Це означає, що збільшення площі на 1% асоціюється приблизно з 0,063% вищою ціною оренди за гектар.

Однак пояснювальна здатність моделі дуже низька: R² = 0,005. Тобто площа сама по собі пояснює лише близько 0,5% варіації ціни оренди за гектар.

Практично цей зв’язок є дуже слабким.

### OLS-модель з контролями

```python
area_control_data = df_clean[["Площа земельної ділянки, ГА", "price_lease_ha", "Кількість пропозицій", "Виставлена вартість"]].dropna()
area_control_data = area_control_data[(area_control_data["Площа земельної ділянки, ГА"] > 0) &
                                      (area_control_data["price_lease_ha"] > 0) &
                                      (area_control_data["Виставлена вартість"] > 0)].copy()

area_control_data["log_area"] = np.log(area_control_data["Площа земельної ділянки, ГА"])
area_control_data["log_price_lease_ha"] = np.log(area_control_data["price_lease_ha"])
area_control_data["log_start_price"] = np.log(area_control_data["Виставлена вартість"])

area_control_model = smf.ols(
    'log_price_lease_ha ~ log_area + Q("Кількість пропозицій") + log_start_price',
    data=area_control_data
).fit()

area_control_results = pd.DataFrame({
    "coefficient": area_control_model.params,
    "p_value": area_control_model.pvalues
})
print(f"R-squared: {area_control_model.rsquared:.3f}")
area_control_results
```

Після додавання контролів за кількістю ставок і стартовою ціною коефіцієнт площі змінює знак і стає від’ємним: β = -0,683.

Модель пояснює близько 47,9% варіації логарифма ціни оренди за гектар (R² = 0,479).

Однак цей негативний коефіцієнт не можна трактувати як чистий економічний ефект площі. Показник `price_lease_ha` розраховується як:

`price_lease_ha = income / area`

тобто площа вже математично входить у знаменник залежної змінної.

Тому сильний негативний коефіцієнт у контрольованій моделі частково може бути наслідком самої конструкції показника.

### Bootstrap-довірчий інтервал

```python
np.random.seed(42)

small_plots = area_data.loc[area_data["area_group"] == "<1 ha", "price_lease_ha"].to_numpy()
large_plots = area_data.loc[area_data["area_group"] == "50+ ha", "price_lease_ha"].to_numpy()

observed_area_difference = np.median(small_plots) - np.median(large_plots)
bootstrap_area_differences = []

for _ in range(1000):
    small_sample = np.random.choice(small_plots, size=len(small_plots), replace=True)
    large_sample = np.random.choice(large_plots, size=len(large_plots), replace=True)
    bootstrap_area_differences.append(np.median(small_sample) - np.median(large_sample))

area_confidence_interval = np.percentile(bootstrap_area_differences, [2.5, 97.5])
print(f"Observed median difference: {observed_area_difference:.2f} UAH per ha")
print(f"95% confidence interval: [{area_confidence_interval[0]:.2f}, {area_confidence_interval[1]:.2f}]")

plt.figure(figsize=(8, 5))
plt.hist(bootstrap_area_differences, bins=30, edgecolor="black")
plt.axvline(0, color="red", linestyle="--", label="Zero difference")
plt.axvline(area_confidence_interval[0], color="orange", linestyle="--", label="95% CI bounds")
plt.axvline(area_confidence_interval[1], color="orange", linestyle="--")
plt.title("Bootstrap median differences: <1 ha vs 50+ ha")
plt.xlabel("Median price per ha difference, UAH")
plt.ylabel("Frequency")
plt.legend()
plt.show()
```

Спостережувана різниця медіанної ціни оренди за гектар між ділянками площею менше 1 га та ділянками 50+ га становить -9 321,64 грн/га.

95% bootstrap-довірчий інтервал становить [-10 964,59; -7 275,05] грн/га і не включає нуль.

Отже, у вибірці ділянки менше 1 га мають нижчу медіанну ціну оренди за гектар, ніж ділянки площею 50+ га.

Цей результат описує відмінність між групами, але не доводить причинного впливу площі на ціну.

### Підсумок розділу 6.2

Зв’язок між площею земельної ділянки та річною ціною оренди за гектар є слабким і нестабільним між різними моделями.

Кореляція Спірмена становить лише 0,088, а проста OLS-модель пояснює близько 0,5% варіації ціни за гектар. Хоча групові медіани та тест Краскела—Волліса показують статистично значущі відмінності між групами площі, сам по собі розмір ділянки має дуже слабку пояснювальну здатність.

Після додавання контролів коефіцієнт площі змінює знак і стає від’ємним. Однак цей результат частково може бути механічним, оскільки площа вже входить у знаменник показника `price_lease_ha`.

Отже, **немає підстав вважати площу земельної ділянки сильним самостійним фактором, що визначає ціну оренди за гектар**. Результат цього розділу слід розглядати як дослідницький та описовий, а не причинний.

Отже, різниця між групами є статистично значущою, але належність до `war_affected` сама по собі має дуже низьку пояснювальну здатність.

### OLS-модель з контролями


```python
war_control_data = df_clean[["war_affected", "year", "price_lease_ha", "Кількість пропозицій", "Площа земельної ділянки, ГА", "Виставлена вартість"]].dropna()
war_control_data = war_control_data[(war_control_data["price_lease_ha"] > 0) &
                                    (war_control_data["Площа земельної ділянки, ГА"] > 0) &
                                    (war_control_data["Виставлена вартість"] > 0)].copy()

war_control_data["log_price_lease_ha"] = np.log(war_control_data["price_lease_ha"])
war_control_data["log_area"] = np.log(war_control_data["Площа земельної ділянки, ГА"])
war_control_data["log_start_price"] = np.log(war_control_data["Виставлена вартість"])
war_control_data["post_2022"] = war_control_data["year"] >= 2022

war_control_model = smf.ols(
    'log_price_lease_ha ~ war_affected + Q("Кількість пропозицій") + log_area + log_start_price + C(year)',
    data=war_control_data
).fit()

war_control_beta = war_control_model.params["war_affected[T.True]"]
war_control_results = pd.DataFrame({
    "war_affected_coefficient": [war_control_beta],
    "p_value": [war_control_model.pvalues["war_affected[T.True]"]],
    "percent_difference": [(np.exp(war_control_beta) - 1) * 100]
})
print(f"R-squared: {war_control_model.rsquared:.3f}")
war_control_results
```

Після контролю кількості ставок, площі земельної ділянки, стартової ціни та року негативна асоціація зберігається.

У контрольованій моделі належність до групи `war_affected` асоціюється приблизно з 20,6% нижчою річною ціною оренди за гектар.

Модель у цілому пояснює близько 49,9% варіації логарифма ціни за гектар (R² ≈ 0,499).

Однак цей результат не можна трактувати як причинний ефект війни, оскільки групи могли систематично відрізнятися ще до 2022 року.

### Дослідницька модель із взаємодією

```python
war_control_data["war_affected_int"] = war_control_data["war_affected"].astype(int)
war_control_data["post_2022_int"] = war_control_data["post_2022"].astype(int)

interaction_model = smf.ols(
    'log_price_lease_ha ~ war_affected_int + C(year) + war_affected_int:post_2022_int + Q("Кількість пропозицій") + log_area + log_start_price',
    data=war_control_data
).fit()

interaction_results = pd.DataFrame({
    "coefficient": interaction_model.params[["war_affected_int", "war_affected_int:post_2022_int"]],
    "p_value": interaction_model.pvalues[["war_affected_int", "war_affected_int:post_2022_int"]]
})
print(f"R-squared: {interaction_model.rsquared:.3f}")
interaction_results
```

Для перевірки того, чи змінилася різниця між групами після 2022 року, було додано interaction між `war_affected` і періодом після 2022 року.

Коефіцієнт взаємодії є позитивним, але статистично незначущим (p ≈ 0,27).

Отже, ця модель не дає достатніх підстав стверджувати, що після 2022 року різниця між двома групами статистично значущо змінилася.

Це не є повноцінним difference-in-differences аналізом і не дозволяє оцінювати причинний ефект війни.

## 6.5 Стартова ціна відносно нормативної грошової оцінки

```python
ratio_data = df_clean[(df_clean["Нормативна грошова оцінка"] > 0) &
                      (df_clean["Виставлена вартість"] > 0)].copy()
ratio_data["start_to_normative"] = (
    ratio_data["Виставлена вартість"] / ratio_data["Нормативна грошова оцінка"]
)

print(f"Rows remaining: {len(ratio_data):,}")
```

### Розрахунок показника

Для аналізу використано показник:

`start_to_normative = стартова ціна / нормативна грошова оцінка`

Він показує, яку частку від нормативної грошової оцінки становить стартова ціна аукціону.

Розподіл цього показника сильно концентрований: 25-й, 50-й і 75-й перцентилі близькі до 0,12, водночас є окремі екстремальні значення до приблизно 12. Тому результати моделей потрібно інтерпретувати обережно.

```python
print(ratio_data["start_to_normative"].describe(
    percentiles=[0.01, 0.05, 0.25, 0.5, 0.75, 0.95, 0.99]
))
print("Unique ratio values:", ratio_data["start_to_normative"].nunique())

plt.figure(figsize=(8, 5))
plt.hist(ratio_data["start_to_normative"], bins=50, edgecolor="black")
plt.xscale("log")
plt.title("Distribution of starting price to normative valuation ratio")
plt.xlabel("Start price / normative valuation (log scale)")
plt.ylabel("Frequency")
plt.show()

ratio_data.nlargest(10, "start_to_normative")[[
    "Виставлена вартість", "Нормативна грошова оцінка", "start_to_normative",
    "Кількість пропозицій", "price_change_pct", "price_lease_ha"
]]
```

### Кореляції Спірмена

```python
from scipy.stats import spearmanr

ratio_spearman_data = ratio_data[["start_to_normative", "Кількість пропозицій", "price_change_pct"]].dropna()
ratio_bids_corr = ratio_spearman_data[["start_to_normative", "Кількість пропозицій"]].corr(method="spearman").iloc[0, 1]
ratio_growth_corr = ratio_spearman_data[["start_to_normative", "price_change_pct"]].corr(method="spearman").iloc[0, 1]
ratio_bids_p = spearmanr(ratio_spearman_data["start_to_normative"], ratio_spearman_data["Кількість пропозицій"]).pvalue
ratio_growth_p = spearmanr(ratio_spearman_data["start_to_normative"], ratio_spearman_data["price_change_pct"]).pvalue

print(f"Spearman correlation: ratio and bids = {ratio_bids_corr:.3f}, p-value = {ratio_bids_p:.3e}")
print(f"Spearman correlation: ratio and price growth = {ratio_growth_corr:.3f}, p-value = {ratio_growth_p:.3e}")
```

Кореляція Спірмена між відношенням стартової ціни до НГО та кількістю ставок становить 0,028, а зі зростанням ціни — 0,020.

Обидва зв’язки статистично значущі, але практично майже нульові.

Отже, вища стартова ціна відносно НГО лише дуже слабко пов’язана з більшою конкуренцією та більшим зростанням ціни.Both Spearman correlations are positive but very weak. Higher starting price relative to normative valuation is associated with only slightly more bids and slightly higher price growth.


### Групування за квартилями показника

```python
ratio_group_data = ratio_data[["start_to_normative", "Кількість пропозицій", "price_change_pct", "price_lease_ha"]].dropna().copy()
ratio_group_data["ratio_quartile"] = pd.qcut(
    ratio_group_data["start_to_normative"], q=4, labels=["Q1", "Q2", "Q3", "Q4"]
)

ratio_summary = ratio_group_data.groupby("ratio_quartile", observed=False).agg(
    auction_count=("start_to_normative", "size"),
    median_start_to_normative=("start_to_normative", "median"),
    median_bids=("Кількість пропозицій", "median"),
    median_price_change_pct=("price_change_pct", "median"),
    median_price_lease_ha=("price_lease_ha", "median")
)
ratio_summary.round(3)
```

```python
plt.figure(figsize=(8, 5))
plt.bar(ratio_summary.index.astype(str), ratio_summary["median_bids"])
plt.title("Median bids by starting-price ratio quartile")
plt.xlabel("Start to normative valuation quartile")
plt.ylabel("Median number of bids")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.bar(ratio_summary.index.astype(str), ratio_summary["median_price_change_pct"])
plt.title("Median price growth by starting-price ratio quartile")
plt.xlabel("Start to normative valuation quartile")
plt.ylabel("Median price growth, %")
plt.show()
```

Медіанні значення не демонструють чіткого монотонного патерну.

Найвищі медіанні значення кількості ставок, зростання ціни та ціни оренди за гектар спостерігаються у третьому квартилі, тоді як у четвертому вони знову знижуються.

Крім того, медіанне значення `start_to_normative` однакове — близько 0,12 — у Q2, Q3 і Q4. Це означає, що квартильне групування тут має обмежену інформативність через сильну концентрацію самого показника.

### Тести Краскела—Волліса

```python
from scipy.stats import kruskal

ratio_bids_groups = [ratio_group_data.loc[ratio_group_data["ratio_quartile"] == group, "Кількість пропозицій"]
                     for group in ["Q1", "Q2", "Q3", "Q4"]]
ratio_growth_groups = [ratio_group_data.loc[ratio_group_data["ratio_quartile"] == group, "price_change_pct"]
                       for group in ["Q1", "Q2", "Q3", "Q4"]]
ratio_bids_test = kruskal(*ratio_bids_groups)
ratio_growth_test = kruskal(*ratio_growth_groups)

print(f"Number of bids: H = {ratio_bids_test.statistic:.2f}, p-value = {ratio_bids_test.pvalue:.3e}")
print(f"Price growth: H = {ratio_growth_test.statistic:.2f}, p-value = {ratio_growth_test.pvalue:.3e}")
```

Тести Краскела—Волліса є статистично значущими як для кількості ставок, так і для зростання ціни між квартильними групами (p < 0,001).

Однак групові медіани не формують чіткого монотонного патерну, тому статистична значущість не означає сильного або послідовного практичного зв’язку.

### Прості OLS-моделі

```python
import statsmodels.formula.api as smf

growth_ratio_model = smf.ols("price_change_pct ~ start_to_normative", data=ratio_group_data).fit()

price_ratio_data = ratio_group_data[ratio_group_data["price_lease_ha"] > 0].copy()
price_ratio_data["log_price_lease_ha"] = np.log(price_ratio_data["price_lease_ha"])
price_ratio_model = smf.ols("log_price_lease_ha ~ start_to_normative", data=price_ratio_data).fit()

ratio_ols_summary = pd.DataFrame({
    "ratio_coefficient": [growth_ratio_model.params.iloc[1], price_ratio_model.params.iloc[1]],
    "p_value": [growth_ratio_model.pvalues.iloc[1], price_ratio_model.pvalues.iloc[1]],
    "R_squared": [growth_ratio_model.rsquared, price_ratio_model.rsquared]
}, index=["price_change_pct", "log(price_lease_ha)"])
ratio_ols_summary
```

У простих OLS-моделях зв’язки залишаються дуже слабкими за пояснювальною здатністю.

Для `price_change_pct` модель має R² = 0,0003, а для `log(price_lease_ha)` — R² = 0,0026.

Це означає, що відношення стартової ціни до НГО саме по собі пояснює практично нульову частку варіації результатів аукціону.

Також напрям коефіцієнта для зростання ціни в OLS є від’ємним, тоді як кореляція Спірмена була слабко позитивною. Така нестабільність напряму додатково свідчить про відсутність стійкого зв’язку.

### Перевірка стійкості до екстремальних значень

```python
ratio_threshold = ratio_group_data["start_to_normative"].quantile(0.99)
trimmed_ratio_data = ratio_group_data[ratio_group_data["start_to_normative"] <= ratio_threshold].copy()

trimmed_growth_model = smf.ols("price_change_pct ~ start_to_normative", data=trimmed_ratio_data).fit()
trimmed_price_data = trimmed_ratio_data[trimmed_ratio_data["price_lease_ha"] > 0].copy()
trimmed_price_data["log_price_lease_ha"] = np.log(trimmed_price_data["price_lease_ha"])
trimmed_price_model = smf.ols("log_price_lease_ha ~ start_to_normative", data=trimmed_price_data).fit()

robustness_summary = pd.DataFrame({
    "coefficient_before": [growth_ratio_model.params.iloc[1], price_ratio_model.params.iloc[1]],
    "coefficient_after": [trimmed_growth_model.params.iloc[1], trimmed_price_model.params.iloc[1]],
    "R_squared_before": [growth_ratio_model.rsquared, price_ratio_model.rsquared],
    "R_squared_after": [trimmed_growth_model.rsquared, trimmed_price_model.rsquared]
}, index=["price_change_pct", "log(price_lease_ha)"])
robustness_summary
```

Після видалення верхнього 1% значень `start_to_normative` напрям коефіцієнтів не змінюється, але їхня величина помітно змінюється.

При цьому R² обох моделей залишається надзвичайно низьким.

Отже, екстремальні значення впливають на оцінку коефіцієнтів, але не створюють практично сильного зв’язку.

### OLS-модель з контролями

```python
ratio_control_data = ratio_data[["start_to_normative", "price_lease_ha", "Кількість пропозицій", "Площа земельної ділянки, ГА", "year"]].dropna()
ratio_control_data = ratio_control_data[(ratio_control_data["price_lease_ha"] > 0) &
                                        (ratio_control_data["Площа земельної ділянки, ГА"] > 0)].copy()
ratio_control_data["log_price_lease_ha"] = np.log(ratio_control_data["price_lease_ha"])
ratio_control_data["log_area"] = np.log(ratio_control_data["Площа земельної ділянки, ГА"])

ratio_control_model = smf.ols(
    'log_price_lease_ha ~ start_to_normative + Q("Кількість пропозицій") + log_area + C(year)',
    data=ratio_control_data
).fit()

ratio_control_results = pd.DataFrame({
    "ratio_coefficient": [ratio_control_model.params["start_to_normative"]],
    "p_value": [ratio_control_model.pvalues["start_to_normative"]],
    "R_squared": [ratio_control_model.rsquared]
})
ratio_control_results
```

Після контролю конкуренції, площі земельної ділянки та року коефіцієнт `start_to_normative` залишається позитивним і статистично значущим.

Коефіцієнт становить приблизно 0,420, а модель у цілому має R² ≈ 0,260.

Водночас цей результат не слід трактувати причинно. Позитивний коефіцієнт може частково відображати особливості формування цінових показників, характеристики земельних ділянок, склад аукціонів та практики оцінювання, які не повністю враховані в моделі.

### Підсумок розділу 6.5

Відношення стартової ціни до нормативної грошової оцінки має статистично значущі, але практично дуже слабкі зв’язки з конкуренцією та ціновими результатами аукціону.

Кореляції Спірмена близькі до нуля, квартильні медіани не показують чіткого монотонного патерну, а прості OLS-моделі пояснюють практично нульову частку варіації.

Крім того, напрям зв’язку зі зростанням ціни відрізняється між Spearman і OLS, що свідчить про нестійкість результату.

Отже, **стартова ціна відносно НГО не є сильним самостійним пояснювальним фактором результатів аукціону**. Контрольована модель показує позитивну асоціацію з ціною за гектар, але цей результат не можна трактувати як причинний.

### OLS-моделі

```python
import statsmodels.formula.api as smf

repeat_ols_data = df_clean[["repeat_auction", "price_lease_ha"]].dropna()
repeat_ols_data = repeat_ols_data[repeat_ols_data["price_lease_ha"] > 0].copy()
repeat_ols_data["log_price_lease_ha"] = np.log(repeat_ols_data["price_lease_ha"])

repeat_model = smf.ols("log_price_lease_ha ~ repeat_auction", data=repeat_ols_data).fit()
repeat_beta = repeat_model.params.iloc[1]

repeat_ols_summary = pd.DataFrame({
    "repeat_auction_coefficient": [repeat_beta],
    "p_value": [repeat_model.pvalues.iloc[1]],
    "R_squared": [repeat_model.rsquared],
    "percent_difference": [(np.exp(repeat_beta) - 1) * 100]
})
repeat_ols_summary
```

```python
repeat_control_data = df_clean[["repeat_auction", "price_lease_ha", "Кількість пропозицій", "Площа земельної ділянки, ГА", "Виставлена вартість", "year"]].dropna()
repeat_control_data = repeat_control_data[(repeat_control_data["price_lease_ha"] > 0) &
                                          (repeat_control_data["Площа земельної ділянки, ГА"] > 0) &
                                          (repeat_control_data["Виставлена вартість"] > 0)].copy()
repeat_control_data["log_price_lease_ha"] = np.log(repeat_control_data["price_lease_ha"])
repeat_control_data["log_area"] = np.log(repeat_control_data["Площа земельної ділянки, ГА"])
repeat_control_data["log_start_price"] = np.log(repeat_control_data["Виставлена вартість"])

repeat_control_model = smf.ols(
    'log_price_lease_ha ~ repeat_auction + Q("Кількість пропозицій") + log_area + log_start_price + C(year)',
    data=repeat_control_data
).fit()

repeat_control_beta = repeat_control_model.params["repeat_auction[T.True]"]
repeat_control_results = pd.DataFrame({
    "repeat_auction_coefficient": [repeat_control_beta],
    "p_value": [repeat_control_model.pvalues["repeat_auction[T.True]"]],
    "R_squared": [repeat_control_model.rsquared],
    "percent_difference": [(np.exp(repeat_control_beta) - 1) * 100]
})
repeat_control_results
```

У простій OLS-моделі повторне виставлення асоціюється приблизно з 30,1% нижчою річною ціною оренди за гектар.

Однак пояснювальна здатність простої моделі невисока: R² ≈ 0,018.

Після контролю кількості ставок, площі земельної ділянки, стартової ціни та року негативна асоціація зменшується, але залишається статистично значущою: повторні аукціони мають приблизно на 12,8% нижчу ціну за гектар.

Контрольована модель має R² ≈ 0,492.

Зменшення оцінки з -30,1% до -12,8% означає, що значна частина початкової різниці пов’язана з іншими характеристиками аукціонів, але негативна асоціація з повторним виставленням зберігається.

### Підсумок розділу 6.6

Повторні аукціони мають нижчу медіанну конкуренцію, нижчу річну ціну оренди за гектар і слабше зростання ціни порівняно з первинними виставленнями.

У простій OLS-моделі повторне виставлення асоціюється приблизно з 30,1% нижчою ціною за гектар, а після контролю конкуренції, площі, стартової ціни та року різниця зменшується приблизно до 12,8%.

## 6.9 Матриця кореляцій

```python
import numpy as np

correlation_columns = [
    "Виставлена вартість", "Дохід", "Кількість пропозицій",
    "Нормативна грошова оцінка", "Площа земельної ділянки, ГА",
    "price_lease_ha", "price_change", "price_change_pct","lease_duration_years_int"]
if "start_to_normative" in df_clean.columns:
    correlation_columns.append("start_to_normative")

correlation_data = df_clean[correlation_columns]
corr = correlation_data.corr(method="spearman")
corr.round(2)
```

```python
plt.figure(figsize=(10, 8))
plt.imshow(corr, cmap="coolwarm", vmin=-1, vmax=1)
plt.colorbar(label="Spearman correlation")
plt.xticks(range(len(corr.columns)), corr.columns, rotation=90)
plt.yticks(range(len(corr.index)), corr.index)

for row in range(len(corr.index)):
    for column in range(len(corr.columns)):
        plt.text(column, row, f"{corr.iloc[row, column]:.2f}", ha="center", va="center")

plt.title("Spearman correlation matrix")
plt.tight_layout()
plt.show()
```

```python
correlation_pairs = corr.where(np.triu(np.ones(corr.shape), k=1).astype(bool)).stack().reset_index()
correlation_pairs.columns = ["variable_1", "variable_2", "correlation"]
correlation_pairs["absolute_correlation"] = correlation_pairs["correlation"].abs()
correlation_pairs["strength"] = pd.cut(
    correlation_pairs["absolute_correlation"],
    bins=[-0.01, 0.4, 0.7, float("inf")],
    labels=["weak", "moderate", "strong"]
)
strongest_correlations = correlation_pairs.sort_values("absolute_correlation", ascending=False).head(15)
strongest_correlations.round(2)
```

- медіанну нормативну грошову оцінку;
- медіанну площу земельної ділянки.

До аналізу включено лише області з щонайменше 30 аукціонами. Після фільтрації залишилося 23 області.

Показники з вираженою асиметрією були логарифмовані, після чого всі ознаки стандартизовано за допомогою `StandardScaler`.

### Метод головних компонент (PCA)

PCA використано для зменшення кількості взаємопов’язаних регіональних показників до двох узагальнених компонент.

Перша компонента, PC1, пояснює 47,1% загальної варіації, друга, PC2, — 25,7%.

Разом PC1 і PC2 пояснюють 72,8% регіональних відмінностей, тому двовимірне представлення зберігає значну частину інформації.

### Інтерпретація компонент

PC1 найбільше пов’язана з:
- нормативною грошовою оцінкою;
- площею земельної ділянки;
- стартовою ціною.

У протилежному напрямі з PC1 пов’язане зростання ціни.

PC2 найбільше пов’язана з:
- кількістю ставок;
- річною ціною оренди за гектар;
- зростанням ціни.

Таким чином, PC1 умовно відображає структурні характеристики земельних лотів, а PC2 — ринкову активність і ціновий результат аукціонів.

### Кластеризація регіонів

Після PCA регіони додатково було згруповано методом `KMeans`.

Для вибору кількості кластерів порівнювали значення silhouette score:

- k = 3 → 0,259;
- k = 4 → 0,252;
- k = 5 → 0,256.

Найвище значення отримано для k = 3, тому для візуалізації використано три кластери.

### Інтерпретація кластерів

Silhouette score близько 0,26 є досить низьким, а результати для k = 3, 4 і 5 дуже близькі між собою.

Це означає, що області не формують чітко відокремлених природних кластерів. Межі між групами є розмитими, а вибір трьох кластерів має радше дослідницький характер.

Тому кластери не слід трактувати як стабільні або природні «типи» регіонів.

```python
regional_pca_data = df_clean[["область", "Кількість пропозицій", "price_lease_ha", "price_change_pct", "Виставлена вартість", "Нормативна грошова оцінка", "Площа земельної ділянки, ГА"]].dropna().copy()

regional_pca_summary = regional_pca_data.groupby("область").agg(
    auction_count=("область", "size"),
    median_price_lease_ha=("price_lease_ha", "median"),
    median_bids=("Кількість пропозицій", "median"),
    median_price_change_pct=("price_change_pct", "median"),
    median_start_price=("Виставлена вартість", "median"),
    median_normative_value=("Нормативна грошова оцінка", "median"),
    median_area=("Площа земельної ділянки, ГА", "median")
)
regional_pca_summary = regional_pca_summary[regional_pca_summary["auction_count"] >= 30].copy()
print(f"Oblasts remaining: {len(regional_pca_summary)}")
print("Missing values:")
print(regional_pca_summary.isna().sum())
print("Skewness:")
print(regional_pca_summary.skew())
```

```python
pca_features = pd.DataFrame(index=regional_pca_summary.index)
pca_features["log_auction_count"] = np.log(regional_pca_summary["auction_count"])
pca_features["log_median_price_lease_ha"] = np.log(regional_pca_summary["median_price_lease_ha"])
pca_features["median_bids"] = regional_pca_summary["median_bids"]
pca_features["log_median_price_change_pct"] = np.log(regional_pca_summary["median_price_change_pct"])
pca_features["log_median_start_price"] = np.log(regional_pca_summary["median_start_price"])
pca_features["log_median_normative_value"] = np.log(regional_pca_summary["median_normative_value"])
pca_features["log_median_area"] = np.log(regional_pca_summary["median_area"])

feature_names = pca_features.columns
```

```python
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
scaled_features = scaler.fit_transform(pca_features)
pca = PCA(n_components=2)
pca_scores = pca.fit_transform(scaled_features)

loadings = pd.DataFrame(
    pca.components_.T,
    index=feature_names,
    columns=["PC1", "PC2"]
)
loadings["abs_PC1"] = loadings["PC1"].abs()
loadings["abs_PC2"] = loadings["PC2"].abs()
loadings.round(3)
```

```python
explained_variance = pd.Series(
    pca.explained_variance_ratio_,
    index=["PC1", "PC2"],
    name="explained_variance_ratio"
)
explained_variance.loc["PC1 + PC2"] = explained_variance.sum()
explained_variance.round(3)
```

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

silhouette_scores = []
for k in [3, 4, 5]:
    labels = KMeans(n_clusters=k, random_state=42, n_init=10).fit_predict(scaled_features)
    silhouette_scores.append([k, silhouette_score(scaled_features, labels)])

silhouette_table = pd.DataFrame(silhouette_scores, columns=["k", "silhouette_score"])
silhouette_table
```

```python
selected_k = silhouette_table.loc[silhouette_table["silhouette_score"].idxmax(), "k"]
regional_pca_summary["cluster"] = KMeans(
    n_clusters=selected_k, random_state=42, n_init=10
).fit_predict(scaled_features)

pca_plot_data = pd.DataFrame(pca_scores, index=regional_pca_summary.index, columns=["PC1", "PC2"])
pca_plot_data["cluster"] = regional_pca_summary["cluster"]
cluster_centroids = pca_plot_data.groupby("cluster")[["PC1", "PC2"]].mean()
pc1_variance = pca.explained_variance_ratio_[0] * 100
pc2_variance = pca.explained_variance_ratio_[1] * 100

plt.figure(figsize=(12, 8))
for cluster in sorted(pca_plot_data["cluster"].unique()):
    cluster_data = pca_plot_data[pca_plot_data["cluster"] == cluster]
    plt.scatter(cluster_data["PC1"], cluster_data["PC2"], s=90, alpha=0.75, label=f"Cluster {cluster}")

plt.scatter(cluster_centroids["PC1"], cluster_centroids["PC2"], s=240, marker="X", color="black", label="Cluster centroids")
plt.axhline(0, color="gray", linewidth=0.8, linestyle="--")
plt.axvline(0, color="gray", linewidth=0.8, linestyle="--")

label_offsets = [(6, 6), (6, -10), (-6, 6), (-6, -10)]
for index, (oblast, row) in enumerate(pca_plot_data.iterrows()):
    offset = label_offsets[index % len(label_offsets)]
    plt.annotate(oblast, (row["PC1"], row["PC2"]), xytext=offset, textcoords="offset points", fontsize=7)

plt.title("Oblasts in PCA space by cluster")
plt.xlabel(f"PC1 ({pc1_variance:.1f}% variance)")
plt.ylabel(f"PC2 ({pc2_variance:.1f}% variance)")
plt.legend(loc="upper left", bbox_to_anchor=(1.02, 1))
plt.tight_layout()
plt.show()
```

```python
cluster_profiles = regional_pca_summary.groupby("cluster").agg(
    oblast_count=("cluster", "size"),
    median_auction_count=("auction_count", "median"),
    median_price_lease_ha=("median_price_lease_ha", "median"),
    median_bids=("median_bids", "median"),
    median_price_change_pct=("median_price_change_pct", "median"),
    median_start_price=("median_start_price", "median"),
    median_normative_value=("median_normative_value", "median"),
    median_area=("median_area", "median")
)
cluster_profiles.round(2)
```

### Підсумок розділу 6.10

PCA показує, що 72,8% регіональної варіації можна описати двома головними компонентами.

Перша компонента переважно характеризує масштаб і вартісні характеристики земельних ділянок, а друга — конкуренцію та цінові результати аукціонів.

Водночас кластерний аналіз не виявив чіткої природної сегментації областей: silhouette score залишається низьким і майже не відрізняється для 3, 4 та 5 кластерів.

Отже, **між регіонами є багатовимірні відмінності, але вони не формують чітких природних кластерів**. PCA є корисним для візуалізації структури регіональних відмінностей, тоді як кластеризацію слід розглядати лише як дослідницький інструмент.
